When error in the result of an algorithm exceeds what conditioning can explain, we call that algorithm unstable

whenever there is addition or subtraction in an algorithm, that is a likely cause instibility in an algorithm

it is possible to restructure algorithms to use mulltiplication and division instead of addition and subtraction, which acn eliminate error issues that arose due to cancellation.

The sensitivity of a problem f(x) depends only on its condition number K, but the sensitivity of an algorithm depends on the condition numbers of all its individual steps

A useful alternative measurement to error asks "Did you answer nearly the right question" instead of "Did you get nearly the right answer". This is called backward error.

Backward error measures the change to the original data that repoduces the result that was found by the algorithm

Imagine a case of evaluating a polynomial near a multiple root. For example, evaluating p(x) = (x-1)^10 at x = 1.0001. 

The trual value is (0.0001)^10 = 10^-40

For the forward error, small rounding errors get amplified massively as we compute the 10th power, so the relative forward error can be several order of magnitude. This, its large/poor

For the backward error, the computed result can be shown to be the exact value of the polynomial p-tilde(x) = (x-1)^10 with coefficients changed by only a very small amount. So, the backward error is small/good.

This happens because we evaluated a polynomial near a point where its derivative is very small (near a multiple root), so: 
1. small input changes cause huge output swings, so the condition number gets large
2. rounding errors are unavoidable when computing (x-1)^10, but they are small in magnitude compared to the problem's sensitivity
3. the algorithm is actually doing well. its computing the exact value of a nearly-identical polynomial, so the backward error is good. the poor forward error isn't the fault of the algorithm, its a propery of the problem itself

--> forward error can be misleading in ill-conditioned problems. if the backward error is still ok, the algorithm might be fine.
