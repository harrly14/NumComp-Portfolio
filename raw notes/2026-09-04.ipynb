Newton's method is often faster than bisection. We will discuss two methods.

Review of bisection: 

Q-linear convergence is when you get a smooth decrease of convergence. R linear convergence just says its bounded about by a q convergent sequence.

Specifying an interval for bisection is often inconvenient

Roots of even degrees are often problematic

Bisection gives a bound on the solution error (because we know the solution is in the interval)

Convergence rate is pretty slow

Examples  of rootfinding (or systems where one stakeholder has the inputs, but a different stakeholder only knows the outputs)
- function is position of a ball after its thrown, we are interested in how to throw it to hit a target. 
- farming soil nutrient cycling. the forward model would be using certain inputs of NPK to soil, get a time series of nutrients stabilizing at an equilibrium. Root-finding application would be having a target stable NPK, and you want to know what initial nutrient inputs would cause you to get there.
- input of data, rootfinding would be finding the interpolant, or the line that goes through all the data

Newton-raphson method: Assume we have a continuous taylor series (this is not always the case, but is for well-behaved functions). As long as x is close to x_0, we can say that the terms in the series past the first derivitive add next to nothing and can be ignored for an approximation.
