Call K-hat the absolute conditioning number. It is defined as f'(x), the derivative of f(x), if f(x) is differentiable. Otherwise, we use the limit definition of derivatives to find K-hat at a certain point. 

Floating point offers relative accuracy, so its more useful to discuss the raltive condition number, K. If f(x) is differentiable, K is the absolute value of f'(x) times x/f. If K is not differentiable, plug in df/dx (or K-hat) for f'(x). Thus, this number could get huge if f'(x) is huge, x is huge, OR f is tiny (this is the most common and shown below). 

```julia
using Plots
f(x) = x - 1; fp(x) = 1
plot(x -> abs(fp(x)) * abs(x)  / abs(f(x)), xlims=(0,2)) # plots relative contioning number vs x for f(x) = x-1
```

The function e^x-1 is well-conditioned, e^x is well conditioned, but x-1 is ill-conditioned for x close to 1. 

Our way of calculating e^x-1 was unstable. Unstable algorithms do something ill-conditioned. In this case, not every step of the way was well conditioned becasue x - 1 is ill-conditioned (as seen above).

Another example: log(1+x). 

I can't be bothered to type the code, but if we plot relative condition number of log(1+x), we can see it see it blows up near 1. So, log(1+x) is unstable near where x = 0 because the problem (the mathematical function) is ill-conditioned when log(x) is close to 1. But, if we a different algorithm that doesn't take that step (doesn't compute log close to 1 in this case), the algorithm could be stable. 

Reliable = well conditioned and stable

Stability: nearly the right answer to nearly the right question. 
Backward stability: exactly the right answer to nearly the right question

Every backward stable algorithm is stable, but not every stable algorithm is backward stable
