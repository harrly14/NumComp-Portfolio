# number

- can use `typeof` to see type of a variable or output
- number types: float64, float32, bigfloat, int64
    - 32-bit floats can be less expensive, but have rounding errors more often, making them more dangerous
- julia lets you implicitly convert for arithmetic and sich via auto promotion

```julia
@show 3 + 3.0
@show 3.0 + 3.0f0
@show 3 + 3.0f0;
```

Note that the last line above promotes to f32, which is potentially risky, because if you add a large number to float, it can over flow and cause issues, e.g. `300000000 + 3.0f0`

division is NOT integer division like in C. Doing `4/3` would output `1.333`. you can do integer division by doing `4 \div 3` and hitting tab after typing `\div` to make the division sign. integer division truncates towards 0.

# functions

functions have many ways to be declared

```julia
function f(x, y, z = 3)
        ...
end
# call
f(big(3.0), 4, 5) # you can specify input types or leave it implicit

# or
g(x,y) = sqrt(x^2 + y^2)

# or
((x,y) -> sqrt(c^2 + y^2))(3,4) # declaration and call
```

sometimes you want to do something with a user-defined function. luckily, you can have functions of functions. 

```julia
function slope_over(a, b, f)
    (f(b) - f(a))  / (b - a)
end

slope_over(0, 1, sin(x)*x) # sin(x)*x is passed in
```

# floating point arithmetic

- when we do `0.1 + 0.2`, we get something like `0.30000000004` because `0.3` doesn't have an exact binary representation in floating point
- the rounding along the way can lead to surprising errors
- however, floating point arithmetic is very precise and well defined, we just aren't always sure what the rounding is going to do
- there exists a largest number "machine epsilon" which represents the largest number we can compute with floating point arithmetic
- the notation of a circle around an operator means to do that with floating point arithmetic (like a plus with a circle around it, etc)
