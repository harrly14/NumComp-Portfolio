## Bisection algorithm
a rootfinding technique that starts with an interval [a,b] containing a root and does not require derivatives. Suppose $f$ is continuous. What is a sufficient condition for $f$ to have a root on [a,b]?

bisection is good if a function is continuous and either always increasing or decreasing. 

```julia
using Roots

g(x) = cos(x) - x

function bisect(f, a, b, tol) # f = function we want to find root of. a,b = interval endpoints, tol = tolerence
    mid = (a + b) / 2
    if abs(b - a) < tol
        return mid
    elseif hasroot(f, a, mid)
        return bisect(f, a, mid, tol) # cut interval in half until you find a root
    else
        return bisect(f, mid, b, tol)
    end
end

x0 = bisect(g, -1, 3, 1e-5)
x0, g(x0)
```

this function can be pretty inefficient and hard to track. we can modify it to (1) append rather than preprend (better for data structure optimization) and (2) uses a loop instead of recursive to bound stack space

```julia
f(x) = cos(x) - x
hasroot(f, a, b) = f(a) * f(b) < 0
function bisect_iter(f, a, b, tol)
    hist = Float64[]
    while abs(b - a) > tol
        mid = (a + b) / 2
        push!(hist, mid)
        if hasroot(f, a, mid)
            b = mid
        else
            a = mid
        end
    end
    hist
end
```

an iterative algorithm is q-linearly convergent if the error shrinks roughly the same fraction with each step. that is, the ratio of the error of step k+1 and the error of step k is less than 1. The smaller this ratio, the faster the convergence.
