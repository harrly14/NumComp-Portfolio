Subtractive cancellation: a loss of accuracy that occurs when two numbers add or subtract to give a result that is much smaller in magnitude. i.e. adding two numbers -1.0012 and 1.00000 that have been rounded to 5 decimal places and getting -0.0012, which removes information in three digits.

The condition number is a ratio of the relative error of the output to the relative error of the input. If f has at least one continuous derivative, this can be simplified to be the ratio of xf'(x) and f(x)

If you consider a function f(x) = cx, where c is an arbitrary constant (c != 0), you can show that the condition number is 1. So, multiplying by a real number leads to the same relative rror in the result as in the data, and does not cause a potential for cancellation like in addition/subtraction

More common functions and their corresponding condition numbers are here: https://fncbook.com/conditioning/#table-condition-functions

We call a problem poorly conditioned or ill-conditioned when K (the condition number) is large. There is no fixed threshold for the term

Root finding is ill-conditioned for quadratics if the first root is much bigger than the distance between the two roots. This generalizes to polynomials of higher degrees. That is, roots of polynomials are ill-conditioned with respect to changes in polynomial coefficients when they are much closer to each other than to the origin.
