sometimes in householder you can subtract some vectors that are very close to eachother. we can instead reflect in the negative axis. for better stability, we should choose the reflection upon either the positive and negative axis that moves x the largest distance - can just look at x1 in the vector to see if we are in the positive plane or negative and then we just send it to the other plane

the process of creating the vectors in householder is backward stable. we may not create a perfectly orthogonal matrix, but if we could work with the refelctors in exact arithmetic, we could.

classical GS had stability issues with applying block projectors. Householder does not have this same issue with reflectors. that is called CompactWY and is what is used by julia when you do `qr()`

Q in CompactWY is huge, but very efficient for its size. this is because there's a distinction between Full QR and reduced QR. 

another way to look at QR:

R^TR = (QR)^TQR = A^TA

Cholesky QR does:
A^TA and then pulls out the upper triangular reprsentation (R). Then Q = A * R^-1. 

this works well for parallel computation because the main work is A^TA which can work on blocks of a matrix
