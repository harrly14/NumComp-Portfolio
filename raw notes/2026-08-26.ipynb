computing $e^x$
the math: $e^x = \sum_{k=0}^{\infty} x^k/k!$

```julia
function myexp(x):
	sum = 1
	for k in 1:100:
		sum += x^k / factorial(big(k))
	end
	return sum
end
myexp(1) - exp(1)
```

this gets accurate to e-16, but is not very efficient because of big() and factorial(). we can do better 

```julia
function myexp(x)
    sum = 0
    term = 1
    k = 1
    while sum + term != sum
        sum += term
        term *= x / k
        k += 1
    end
    sum
end
myexp(1) - exp(1)
```

This algorithm's approximation runs much faster by ending the while loop when the term gets smaller than machine epsilon (sum + tiny term == sum). This is less expensive because you don't use factorial or big(). This also get e-16 precision.

This function also brings up a good point about designing approximation algorithms, which is that you should let the math tell you when to stop. In the first function, it goes through the loop 100 times no matter what, but in the second function the loop ends once the result is not meaningfully changing creating a dynamic stopping point that cuts out unused steps

More examples were shown with e^x and e^x - 1, but the key points I got were: 
- conditioning is a property of math, while stability is a property of the algorithm. therefore, if your algorithm/function is not stable, you can change it to make it more stable, but if the function isn't conditioned, you can't fix that with a different algorithm (because its a property of the ideal function f, not your function f-tilde)
- subtracting two very simiilar floating point numbers destroys relative accuracy by cancelling leading digits and leaving only the noice left over
    - e.g. f(x) = x - 1, g(x) = f(x) - 1. The rounding happens in step 1 (f(x)), but the loss of precisions happens in step 2 (g(x))
    - when an algorithm leads to cancellation like this, you should try to restructure the algorithm, not just get more precise data types

Reliable = well conditioned and stable
