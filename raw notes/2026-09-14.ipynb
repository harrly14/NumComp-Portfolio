lin alg terminology review:
- range of a matrix is the space spanned by its cols
- nullspace of a matrix A is the space of vectors x such that Ax = 0
- rank of A is dimension of its range
- a matrix is full range if the null space of A or A transpose is empty. equivalently, all calus of A or A transpose are linearly independent
- a nonsingular/invertible matrix is a square matrix of full rank
