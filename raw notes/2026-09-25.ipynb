we can calculate a whole row of the R matrix in Gram Schmidt once we have its first entry. we can modify the algo to something called right looking Gram Schmidt to do the whole row once we know the first entry.

when to use classical vs modified gram schmidt?

classical is unstable, but you don't need to know all vectors in advance so for something like streaming where we get vectors one at a time and want to do this algorithm on the vectors as we receive them

modified needs to be right looking to be efficient, but its less unstable.


this is an example of an algorithm that is used heavily in certain applications but isn't the right tool for the job always. householder QR which we will learn about shortly is better in some scenarios

householder QR:
uses orthogonal transformations (reflectors) to triangularize rather than projectors like in Gram Schmidt

flip the problem on its head: look at Q'A = R rather than A = QR

Take a matrix A that is full of arbitrary values. Construct a matrix Q1 which is a reflector such that Q1 * A introduces all zeroes below the diagonal in the first column. Q's are all orthogonal matrices. Continue with Q2 through to Qn until you have all zeroes below the diagonal - that's R. Q's are m by m and there are n of them (if A is m by n)
