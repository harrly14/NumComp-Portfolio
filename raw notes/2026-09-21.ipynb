naive gram shmidt algorithm:

```python
function gram_schmidt_naive(A)
    m, n = size(A)
    Q = zeros(m, n)
    R = zeros(n, n)
    for j in 1:n
        v = A[:,j]
        for k in 1:j-1
            r = Q[:,k]' * v
            v -= Q[:,k] * r
            R[k,j] = r
        end
        R[j,j] = norm(v)
        Q[:,j] = v / R[j,j]
    end
    Q, R
end
```

if A is 100x10, Q is 100x10 and R is 10x10 because QR must be the shape of A, so these shapes produce the shape of A when multiplied together.

Q is not orthogonal because its not square, but its columns are orthonormal

every full rank matrix that is at least as tall as it is wide has a unique QR factorization (if you apply the constraint that each element of R along the main diagonal must be positive)

how accurate is this naive algorithm?

as we increase m (the dimensionality of a polynomial), our error gets really bad so Q becomes nothing like an orthogonal matrix and is useless

another algorithm called gram_schmidt_classical does more computations in parrallel but gets worse error

cost is m^3, which is typical of factorizations
